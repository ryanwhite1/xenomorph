```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mc
```

```python
grid = np.linspace(-4, 4, 100)
x, y = np.meshgrid(grid, grid)

gaussian = np.exp(-0.5 * (x**2 + y**2)) / (2 * np.pi)
```

```python
fig, ax = plt.subplots()

NORM = mc.PowerNorm(0.35, 1e-4, 1)
NORM = mc.TwoSlopeNorm(0.015, 1e-4, 1)
mesh = ax.pcolormesh(x, y, gaussian, norm=NORM)

# mesh = ax.pcolormesh(x, y, gaussian, norm='log', vmin=1e-4, vmax=1)
fig.colorbar(mesh, ticks=[0.0001, 0.001, 0.01, 0.1, 0.2, 0.4, 0.6, 0.8, 1], format="%1.6s")
```
